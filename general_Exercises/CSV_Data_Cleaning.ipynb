```python
#set up for radial chart
column1="Department"
counts = df[column1].value_counts()
labels = counts.index.tolist()
values = counts.values
N=len(labels)
angles = np.linspace(0,2*np.pi,N,endpoint=False)
fig, ax = plt.subplots(subplot_kw={"projection":"polar"})
bars = ax.bar(angles,values)
ax.set_xticks(angles)
ax.set_xticklabels(labels)
for label, angle in zip(ax.get_xticklabels(),angles):
    label.set_rotation(np.degrees(angle))
plt.title("Radial Chart of Categories")
plt.show()
```

```python
#set up for radial chart, excluding cetain values
column1="Department"
counts1 = df[column1].value_counts()
labels = counts1.index.tolist()
values = counts1.values
N=len(labels)
angles = np.linspace(0,2*np.pi,N,endpoint=False)
exclude_values = ["Drawings and Prints","The Libraries","European Sculpture and Decorative Arts"]
filtered_df= df[~df["Department"].isin(exclude_values)]
counts1= filtered_df["Department"].value_counts().sort_values(ascending=False)
labels=counts1.index.tolist()
values=counts1.values
N=len(labels)
angles =np.linspace(0,2*np.pi,N, endpoint=False)
width=(2*np.pi)/ N*0.8
fig, ax = plt.subplots(subplot_kw={"projection":"polar"})
ax.bar(angles,values,width=width)
ax.set_xticks(angles)
ax.set_xticklabels(labels)
plt.title("radial")
plt.show()
```

```python
x_col="Department"
y_col="Period"

x_labels = df[x_col].astype("category").cat.categories
y_labels = df[y_col].astype("category").cat.categories

x_codes = df[x_col].astype("category").cat.codes
y_code = df[y_col].astype("category").cat.codes

plt.scatter(x_codes,y_code)
plt.xticks(range(len(x_labels)),x_labels, rotation=45)
plt.yticks(range(len(y_labels)),y_labels)

plt.xlabel(x_col)
plt.ylabel(y_col)
plt.title("categorical scatterplot")

#export as svg
plt.savefig("plot.svg", format ="svg", bbox_inches="tight")
plt.show()
```

```python
#load and prepare data
col_a,col_b = 'Period','Department'
#group and count the pairings
pairings = df.groupby([col_a,col_b]).size().reset_index(name='count')
#create scatterplot
plt.figure(figsize=(10,10))
#s defines the size of the points based on frequency count   
plt.scatter(pairings[col_a], pairings[col_b],
            s=pairings['count']*5, #adjust number to change bubble size
            alpha=0.5,
            edgecolors="w",
            linewidth=2)
plt.xlabel('categories in {col_a}')
plt.ylabel('categories in {col_b}')
plt.title("categorical scatterplot")
plt.grid(True, linestyle='--',alpha=0.2)

pairings.to_csv('pairings_frequencies.csv',index=False)


```

```python
#load and prepare data
target_a='Asian Art'
filtered_df = df[df['Department'] == target_a]

counts = filtered_df['Period'].value_counts().reset_index()
counts.columns = ['PeriodCategory','Frequency']
plt.figure(figsize=(10,6))

plt.scatter(counts['PeriodCategory'], counts['Frequency'],
            s=counts['Frequency']*5, #adjust number to change bubble size
            alpha=0.5,

            linewidth=2)
plt.xlabel('categories  1')
plt.ylabel('categories 2')
plt.title("categorical scatterplot")
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()

```
