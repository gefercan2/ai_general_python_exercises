```python
#set up to save as md file
#check in virtual enviroment list of requirements
import docling
import pypdf
from PIL import Image
import docx

print("All docling dependencies loaded ok")
```

```python
from docling.document_converter import DocumentConverter

converter = DocumentConverter()

# Replace this path with an actual PDF you have
#result = converter.convert("path/to/your/file.pdf")
result = converter.convert("/home/german1/Desktop/art1/pdfs/iiae060.pdf")

# Print the result as markdown
print(result.document.export_to_markdown())
```

```python
import json

json_output = result.document.export_to_dict()

with open("output.json", "w") as f:
    json.dump(json_output, f, indent=2)

print("Saved to output.json")
```
